```python
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV, ShuffleSplit # cross validaion set
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor

def find_best_model_using_gridsearchcv(X, y):
    # Define the models and their respective parameter grids
    algos = {
        # 'linear_regression': {
        #     'model': LinearRegression(),
        #     'params': {'fit_intercept': [True, False]}
        # },
        # 'lasso': {
        #     'model': Lasso(),
        #     'params': {
        #         'alpha': [1, 2],
        #         'selection': ['random', 'cyclic'],
        #         'max_iter': [1000, 2000],
        #         'tol': [1e-4, 1e-3],
        #         'warm_start': [True, False]
        #     }
        # },
        'decision_tree': {
            'model': DecisionTreeRegressor(),
            'params': {
                'criterion': ['squared_error', 'friedman_mse'],
                'splitter': ['best', 'random'],
                'max_depth': [None, 10, 20],
                'min_samples_split': [2, 5],
                'min_samples_leaf': [1, 2],
                'max_features': [None, 'sqrt', 'log2'],
                'max_leaf_nodes': [None, 10, 20],
                'random_state': [42]
            }
        },
        'random_forest': {
            'model': RandomForestRegressor(),
            'params': {
                'n_estimators': [50, 100],
                'max_depth': [None, 10, 20],
                'min_samples_split': [2, 5]
            }
        },
        # 'svr': {
        #     'model': SVR(),
        #     'params': {
        #         'C': [0.1, 1, 10],
        #         'kernel': ['linear', 'rbf'],
        #         'gamma': ['scale', 'auto']
        #     }
        # },
        'gradient_boosting': {
            'model': GradientBoostingRegressor(),
            'params': {
                'n_estimators': [100, 200],
                'learning_rate': [0.1, 0.01],
                'max_depth': [3, 5]
            }
        },
        'xgboost': {
            'model': XGBRegressor(),
            'params': {
                'n_estimators': [100, 200],
                'learning_rate': [0.1, 0.01],
                'max_depth': [3, 5]
            }
        },
        # 'knn': {
        #     'model': KNeighborsRegressor(),
        #     'params': {
        #         'n_neighbors': [5, 10],
        #         'weights': ['uniform', 'distance'],
        #         'algorithm': ['auto', 'ball_tree', 'kd_tree']
        #     }
        # }
    }

    scores = []
    # ShuffleSplit cross-validation setup
    cv = ShuffleSplit(n_splits=5, test_size=0.2, random_state=0)

    # Iterate over each algorithm and apply GridSearchCV
    for algo_name, config in algos.items():
        gs = GridSearchCV(config['model'], config['params'], cv=cv, return_train_score=False)
        gs.fit(X, y)

        # Collecting the best score and parameters for each algorithm
        scores.append({
            'model': algo_name,
            'best_score': gs.best_score_,
            'best_params': gs.best_params_
        })

    # Return a DataFrame containing the best scores and parameters
    return pd.DataFrame(scores, columns=['model', 'best_score', 'best_params'])

# Example usage
# X and y are your features and target variables
find_best_model_using_gridsearchcv(X, y)
```

```python
import numpy as np

def predict_price(model, bathrooms, bedrooms, built_year, garage_space, kitchens, sqft, property_type, city):    
    # Create an input array with zeros, matching the number of columns in X
    x = np.zeros(len(X.columns))
    
    # Set values for numerical features
    x[0] = bathrooms
    x[1] = bedrooms
    x[2] = built_year
    x[3] = garage_space
    x[4] = kitchens
    x[5] = sqft
    
    # Set one-hot encoded values for property type
    property_type_column = f"Property Type_{property_type}"
    if property_type_column in X.columns:
        loc_index = np.where(X.columns == property_type_column)[0][0]
        x[loc_index] = 1

    # Set one-hot encoded values for city
    city_column = f"City_{city}"
    if city_column in X.columns:
        loc_index = np.where(X.columns == city_column)[0][0]
        x[loc_index] = 1

    # Predict the price using the provided model
    return model.predict([x])[0]
```
